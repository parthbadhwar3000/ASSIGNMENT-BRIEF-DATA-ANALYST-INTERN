```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
co=pd.read_csv('customer_orders.csv')
pm=pd.read_csv('payments.csv')
```

```python
co.head()
```

```python
pm.head()
```

```python
co['order_status'].value_counts()
```

```python
co['order_year']=co['order_date'].str.split('-').str[0]
```

```python
co['order_month']=co['order_date'].str.split('-').str[1]
co['order_day']=co['order_date'].str.split('-').str[2]
```

```python
co.head()
```

```python
pm.head()
```

```python
pm['payment_year']=pm['payment_date'].str.split('-').str[0]
pm['payment_month']=pm['payment_date'].str.split('-').str[1]
pm['payment_day']=pm['payment_date'].str.split('-').str[2]
```

```python
pm.head()
```

```python
pay_count=pm['payment_status'].value_counts()
pay_count
```

## Data visualization



```python
pay_count.plot.pie(y=pm['payment_status'],figsize=(10,10),autopct='%1.1f%%')
```

```python
co['order_status'].value_counts().plot.pie(y=co['order_status'],figsize=[13,13],autopct='%1.1f%%')
```
